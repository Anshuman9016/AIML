```python
import seaborn as sn
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv('iris.csv')
```

```python
plt.figure(figsize=(9, 7))
plt.plot(df['species'], df['sepal_length'], marker='^', linestyle='-', color='red')
plt.title('Line Plot: Species vs Sepal Length')
plt.xlabel('Species')
plt.ylabel('Sepal Length (cm)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(9, 7))
sn.barplot(x='species', y='sepal_length', data=df,palette=['green'])
plt.title('Bar Plot: Species vs Sepal Length')
plt.xlabel('Species')
plt.ylabel('Sepal Length (cm)')
plt.show()
```

```python
plt.figure(figsize=(9, 7))
plt.scatter(df['species'], df['sepal_length'],marker = '1', color='darkblue')
plt.title('Scatter Plot: Species vs Sepal Length')
plt.xlabel('Species')
plt.ylabel('Sepal Length (cm)')
plt.show()
```

```python
plt.figure(figsize=(9, 7))
sn.boxplot(x='species', y='sepal_length', data=df)
plt.title('Box Plot: Sepal Length by Species')
plt.xlabel('Species')
plt.ylabel('Sepal Length (cm)')
plt.show()
```

```python
plt.figure(figsize=(9, 7))
plt.hist(df['sepal_length'], bins=10, color='orange', edgecolor='b')
plt.title('Histogram: Sepal Length Distribution')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.show()
```

```python
corr_matrix = df.drop(columns='species').corr()
plt.figure(figsize=(8, 6))
sn.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Heatmap: Correlation Matrix of Iris Features')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sn.pairplot(df, hue='species', markers=["o", "s", "D"])
plt.suptitle('Pair Plot: Iris Dataset', y=1.02)
plt.show()
```
