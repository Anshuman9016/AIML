```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame(pd.read_csv('train.csv'))

print(df.head())

plt.figure(figsize=(10, 6))
sns.histplot(df['Pclass'], color='red', label='Pclass', kde=True)
sns.histplot(df['Age'], color='blue', label='Age', kde=True)
sns.histplot(df['Fare'], color='green', label='Fare', kde=True)
plt.legend()
plt.title('Histograms of Pclass, Age, and Fare')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.subplot(1, 3, 1)
sns.boxplot(y=df['Pclass'], color='yellow')
plt.title('Boxplot of Pclass')

plt.subplot(1, 3, 2)
sns.boxplot(y=df['Age'], color='violet')
plt.title('Boxplot of Age')

plt.subplot(1, 3, 3)
sns.boxplot(y=df['Fare'], color='blue')
plt.title('Boxplot of Fare')
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Age', y='Fare', hue='Pclass', data=df, palette='viridis')
plt.title('Scatter Plot: Age vs Fare by Pclass')
plt.show()
```

```python
sns.pairplot(df[['Age', 'Fare', 'Pclass', 'SibSp', 'Parch']], hue='Pclass', palette='bright')
plt.suptitle('Pair Plot of Numerical Features', y=1.02)
plt.show()
```
